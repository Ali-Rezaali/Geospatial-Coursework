# Unsupervised Clustering Exploration
This notebook explores K-Means and BIRCH clustering on the LiDAR point cloud.
We use only the Z-coordinate as a baseline feature to demonstrate the limitations of unsupervised methods on this dataset.

```python
import sys
sys.path.append('../src')
from unsupervised_classification import run_kmeans, run_birch
import laspy
import numpy as np
from sklearn.preprocessing import MinMaxScaler
from collections import Counter

las = laspy.read('../data/sample_points/nonground_training.las')
points = np.vstack([las.X, las.Y, las.Z]).transpose()
features = points[:, 2].reshape(-1, 1)
features = MinMaxScaler().fit_transform(features)

kmeans_labels = run_kmeans(features)
print("K-Means Distribution:", Counter(kmeans_labels))

birch_labels = run_birch(features)
print("BIRCH Distribution:", Counter(birch_labels))
```
